```python
import os
from skimage.measure import regionprops, label
from skimage.morphology import remove_small_objects
import pandas as pd
import numpy as np
import PIL
import seaborn as sns
import matplotlib.pyplot as plt
from skimage import io
PIL.Image.MAX_IMAGE_PIXELS = None
```

```python
rdir = '/srv/data/michielvc/data/other_projects/christian/final_data/segments/'
vsig4_dir = os.path.join(rdir, 'kcs/vsig4/')
f480_dir = os.path.join(rdir, 'kcs/f480/')
files = os.listdir(vsig4_dir)
wdir = '/srv/data/michielvc/data/other_projects/christian/final_data/segments/kcs/f480_subset/'
```

```python
for file in files:
    print(file)
    # read in the images & remove small spots 
    kc_vsig4 = io.imread(os.path.join(vsig4_dir, file)).astype(bool)
    kc_f480 = io.imread(os.path.join(f480_dir, file)) .astype(bool)
    kc_f480 = remove_small_objects(kc_f480, min_size=150)
    kc_vsig4 = remove_small_objects(kc_vsig4, min_size=150)

    # Vsig4+ F4/80 segments are those of which more than 7% of the resp. areas are covered with Vsig4 signal 
    labeled_kc_f480 = label(kc_f480)
    props = regionprops(labeled_kc_f480, intensity_image=kc_vsig4)
    vsig_proportion = np.array([p.mean_intensity for p in props])
    lbs = np.array([p.label for p in props])

    vsig4_thresh = 0.07
    vsig4_plus = vsig_proportion > vsig4_thresh
    vsig4_minus = vsig_proportion <= vsig4_thresh

    vsig4_plus_labels = lbs[vsig4_plus]
    vsig4_minus_labels = lbs[vsig4_minus]

    kc_f480_vsig4_plus = np.isin(labeled_kc_f480, vsig4_plus_labels)
    kc_f480_vsig4_minus = np.isin(labeled_kc_f480, vsig4_minus_labels)
    both = (kc_f480_vsig4_plus * 1) + (kc_f480_vsig4_minus * 2)
    io.imsave(os.path.join(wdir, file), both.astype(np.uint8))
```
